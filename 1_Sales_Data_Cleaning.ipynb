```python
import pandas as pd
import numpy as np

data = {
    'Customer': ['Ahmed', 'Mona', 'Ali', 'Sara', 'Ahmed', 'Omar', None, 'Mona'],
    'Product': ['Laptop', 'Phone', 'Tablet', 'Laptop', 'Laptop', 'Mouse', 'Phone', 'Phone'],
    'Quantity': [2, 1, 3, 1, 2, None, 2, 1],
    'Price': [15000, 8000, 5000, 15000, 15000, 500, 8000, 8000],
    'City': ['Alexandria', 'Cairo', 'Giza', 'Alexandria', 'Alexandria', 'Cairo', None, 'Cairo']
}

df = pd.DataFrame(data)

df.to_csv('sales_data.csv', index=False)

df.head()
```

```python
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
df.describe(include='all')
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Fill missing customer names
df['Customer'] = df['Customer'].fillna('Unknown')

# Fill missing cities
df['City'] = df['City'].fillna('Unknown')

# Fill missing quantities with median
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())

# Convert quantity to integer
df['Quantity'] = df['Quantity'].astype(int)

# Remove extra spaces from text columns
for col in ['Customer', 'Product', 'City']:
    df[col] = df[col].str.strip()

# Create Total Sales column
df['Total_Sales'] = df['Quantity'] * df['Price']

df.head()
```

```python
print("Missing Values After Cleaning:")
print(df.isnull().sum())

print("\nDuplicates After Cleaning:")
print(df.duplicated().sum())

print("\nFinal Dataset:")
display(df)
```

```python
df.to_csv('cleaned_sales_data.csv', index=False)

df.to_excel('cleaned_sales_data.xlsx', index=False)

print("Files saved successfully!")
```
